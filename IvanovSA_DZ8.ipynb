```python
import requests

# Отправка GET-запроса к /posts
response = requests.get('https://jsonplaceholder.typicode.com/posts')

# Проверка успешности запроса
if response.status_code == 200:
    posts = response.json()  # Извлечение данных в JSON

    # Вывод заголовков и тел первых пяти постов
    for post in posts[:5]:
        print(f"Заголовок: {post['title']}")
        print(f"Тело: {post['body']}")
        print()  # Пустая строка, разделение ответа
else:
    print(f"Ошибка: {response.status_code}")
```

```python
import requests

API_KEY = '5f3cdb0de28d4aa980e113954261411'
BASE_URL = 'https://api.weatherapi.com/v1/current.json'
#BASE_URL = 'https://api.openweathermap.org/data/2.5/weather' - не доступен

# Ввод названия города от пользователя
city = input("Введите название города: ").strip()

# Передаваемые параметры в запрос, ответы на русском языке
params = {
    'key': API_KEY,
    'q': city,
    'lang': 'ru'
}
# Отправка GET-запроса к API
response = requests.get(BASE_URL, params=params, timeout=10)

# Проверка успешности запроса
if response.status_code == 200:
    data = response.json()
    # Извлечение температуры и описания погоды
    city_ = data['location']['name'] # для вывода на случай ошибки ввода города
    temperature = data['current']['temp_c']
    weather_description = data['current']['condition']['text']
    weather_description_wind = data ['current']['wind_dir']
    last_weather_observation = data ['current']['last_updated']

    # Вывод результатов
    print("="*80)
    print(f"Текущая температура в городе {city_}: {temperature}°C")
    print(f"Описание погоды: {weather_description}, направление ветра по 16-румбовому компасу - < {weather_description_wind} >")
    print(f"Дата и время крайнего наблюдения: {last_weather_observation}")
else:
    print(f"Ошибка: {response.status_code}. Проверьте название города или ваш API-ключ.")
```
